  Creamos una función auxiliar para dar formato a los datos y mostrar las matrices o vectores alineados y legibles en la pantalla:

<p>
  <font size="3" color="tomato"><b>def ver(m):</b></font>
  <font size="3"> Define la función que recibe un arreglo de NumPy (matriz o vector) para imprimirlo de forma limpia.</font><br>
  
  <font size="3" color="tomato"><b>if m.ndim == 1:</b></font>
  <font size="3"> Condición que evalúa si el arreglo tiene una sola dimensión (Vector).</font><br>
  
  <font size="3" color="tomato"><b>print("[ " + "  ".join(...) + " ]\n")</b></font>
  <font size="3"> Une los elementos del vector en una sola línea separada por espacios, limitando el formato a 2 decimales.</font><br>
  
  <font size="3" color="tomato"><b>else:</b></font>
  <font size="3"> Bloque que se ejecuta si el arreglo es bidimensional (Matriz).</font><br>
  
  <font size="3" color="tomato"><b>for fila in m:</b></font>
  <font size="3"> Ciclo iterativo que recorre y procesa la matriz de forma secuencial, fila por fila.</font><br>
  
  <font size="3" color="tomato"><b>print("[ " + "  ".join(...) + " ]")</b></font>
  <font size="3"> Imprime cada fila de la matriz de manera ordenada, formateando cada probabilidad a 2 decimales para mantener la alineación.</font>
</p>

```python
def ver(m):
    # Vector
    if m.ndim == 1:
        print("[ " + "  ".join(f"{x:.2f}" for x in m) + " ]\n")
    # Matriz
    else:
        for fila in m:
            print("[ " + "  ".join(f"{x:.2f}" for x in fila) + " ]")
        print() # Salto de línea al final
```

<h1><font>¿Cuál es la probabilidad de que el ratón, iniciando  en la casilla 0, alcance la comida?</font></h1>
<p><font size="3"><b>Planteamiento Analitico</b><br><br>
  Para resolver esto, utilizamos el analisis de primeras visitas. <br>
  Definimos <i>u<sub>i</sub></i> como la probabilidad de que el raton, iniciando en la casilla <i>i</i>, alcance la comida (casilla 7) antes de la trampa electrica (casilla 8).
</font></p><br>
<p><font size="3">Estados absorbentes: u<sub>7</sub> = 1 (Comida) y u<sub>8</sub> = 0 (Trampa electrica).</font></p>
<p><font size="3">Estados transitorios: El raton elige de manera uniforme entre las puertas abiertas de cada habitacion. Asi planteamos el sistema de ecuaciones de balance:</font></p>

<p>
  <font size="3" color="orange">1. u<sub>0</sub> = 1/2 u<sub>1</sub> + 1/2 u<sub>2</sub></font><br>
  <font size="3" color="green">2. u<sub>1</sub> = 1/3 u<sub>0</sub> + 1/3 u<sub>3</sub> + 1/3 (1)</font><br>
  <font size="3" color="blue">3. u<sub>2</sub> = 1/3 u<sub>0</sub> + 1/3 u<sub>3</sub> + 1/3 (0)</font><br>
  <font size="3" color="brown">4. u<sub>3</sub> = 1/4 u<sub>1</sub> + 1/4 u<sub>2</sub> + 1/4 u<sub>4</sub> + 1/4 u<sub>5</sub></font><br>
  <font size="3" color="red">5. u<sub>4</sub> = 1/3 u<sub>3</sub> + 1/3 u<sub>6</sub> + 1/3 (1)</font><br>
  <font size="3" color="red">6. u<sub>5</sub> = 1/3 u<sub>3</sub> + 1/3 u<sub>6</sub> + 1/3 (0)</font><br>
  <font size="3" color="purple">7. u<sub>6</sub> = 1/2 u<sub>4</sub> + 1/2 u<sub>5</sub></font>
</p>

<p><font size="3">Solucion:<br>
  Si sumamos las ecuaciones 2 y 3 para simplificar el bloque superior:<br>
  <font size="3" color="green">u<sub>1</sub> + u<sub>2</sub> = 2/3 u<sub>0</sub> + 2/3 u<sub>3</sub> + 1/3</font><br>
  Por la ecuacion 1, sabemos que <font size="3" color="orange">2 u<sub>0</sub> = u<sub>1</sub> + u<sub>2</sub></font>. Sustituyendo:<br>
  <font size="3" color="orange">2 u<sub>0</sub> = 2/3 u<sub>0</sub> + 2/3 u<sub>3</sub> + 1/3 &rArr; 4/3 u<sub>0</sub> = 2/3 u<sub>3</sub> + 1/3</font><br>
  Multiplicando toda la expresion por 3/2 para despejar u<sub>3</sub>:<br>
  <font size="3" color="brown">2 u<sub>0</sub> = u<sub>3</sub> + 1/2 &rArr; u<sub>3</sub> = 2 u<sub>0</sub> - 1/2</font>
</font></p>

<p><font size="3">
  Ahora de manera analoga, sumamos las ecuaciones 5 y 6 para el bloque inferior:<br>
  <font size="3" color="red">u<sub>4</sub> + u<sub>5</sub> = 2/3 u<sub>3</sub> + 2/3 u<sub>6</sub> + 1/3</font><br>
  Por la ecuacion 7, sabemos que <font size="3" color="purple">2 u<sub>6</sub> = u<sub>4</sub> + u<sub>5</sub></font>. Sustituyendo:<br>
  <font size="3" color="purple">2 u<sub>6</sub> = 2/3 u<sub>3</sub> + 2/3 u<sub>6</sub> + 1/3 &rArr; 4/3 u<sub>6</sub> = 2/3 u<sub>3</sub> + 1/3</font><br>
  Multiplicando nuevamente por 3/2 para despejar u<sub>3</sub>:<br>
  <font size="3" color="brown">2 u<sub>6</sub> = u<sub>3</sub> + 1/2 &rArr; u<sub>3</sub> = 2 u<sub>6</sub> - 1/2</font>
</font></p>

<p><font size="3">
  Al igualar ambas expresiones de u<sub>3</sub>, por la simetria del laberinto se obtiene de forma directa que <font size="3" color="red">u<sub>0</sub> = u<sub>6</sub>.</font><br>
  Sustituyendo los pares (<font size="3" color="orange">u<sub>1</sub> + u<sub>2</sub> = 2 u<sub>0</sub></font>) y (<font size="3" color="purple">u<sub>4</sub> + u<sub>5</sub> = 2 u<sub>6</sub> = 2 u<sub>0</sub></font>) en la ecuacion central 4:<br>
  <font size="3" color="brown">u<sub>3</sub> = 1/4 (2 u<sub>0</sub>) + 1/4 (2 u<sub>0</sub>) = 1/2 u<sub>0</sub> + 1/2 u<sub>0</sub></font><br>
  <font size="3" color="red">u<sub>3</sub> = u<sub>0</sub>.</font>
</font></p>

<p><font size="3">
  Se igualan las relaciones de u<sub>3</sub> obtenidas en los pasos anteriores:<br>
  <font size="3" color="orange">u<sub>0</sub> = 2 u<sub>0</sub> - 1/2<br>
  1/2 = 2 u<sub>0</sub> - u<sub>0</sub><br>
  1/2 = u<sub>0</sub></font>
</font></p>

<p><font size="3">
  <font color="red">u<sub>0</sub> = 1/2</font><br>
  La probabilidad analitica de que el raton alcance la comida desde la casilla 0 es exactamente <font color="red">0.5</font> (<font color="red">50%</font>).
</font></p>

<br><p><font size="3">
  <b>Proceso en Python</b><br>
  Escribimos la estructura matemática del laberinto usando arreglos de NumPy:
</font></p>

<p>
  <font size="3" color="green"><b>A = np.array([...])</b></font>
  <font size="3"> Matriz con todas las probabilidades de moverse entre las casillas del laberinto.</font><br>
  <font size="3" color="green"><b>b = np.array([...])</b></font>
  <font size="3"> Vector con las probabilidades de pasar directo a la comida desde cada posición.</font>
</p>

```python
# Matriz A de Probabilidades de transicion
A = np.array([[0,   1/2, 1/2, 0,   0,   0,   0  ],
              [1/3, 0,   0,   1/3, 0,   0,   0  ],
              [1/3, 0,   0,   1/3, 0,   0,   0  ],
              [0,   1/4, 1/4, 0,   1/4, 1/4, 0  ],
              [0,   0,   0,   1/3, 0,   0,   1/3],
              [0,   0,   0,   1/3, 0,   0,   1/3],
              [0,   0,   0,   0,   1/2, 1/2, 0  ]])

# Vector b de Probabilidades de exito
b = np.array([0, 1/3, 0, 0, 1/3, 0, 0])
```

```python
print("A")
ver(A)

print("b")
ver(b)
```

<p><font size="3">
  Planteamos el sistema matricial de balance para calcular los valores exactos de éxito de cada casilla:
</font></p>

<p>
  <font size="3" color="green"><b>I = np.eye(7)</b></font>
  <font size="3"> Genera la matriz identidad necesaria para hacer el despeje algebraico.</font><br>
  <font size="3" color="green"><b>M = I - A</b></font>
  <font size="3"> Resta la matriz de transición a la identidad para agrupar las incógnitas.</font><br>
  <font size="3" color="green"><b>vr = np.linalg.solve(M, b)</b></font>
  <font size="3"> Resuelve el sistema de ecuaciones para obtener el valor real exacto de cada estado.</font><br>
  <font size="3" color="green"><b>print(...)</b></font>
  <font size="3"> Muestra en pantalla el resultado analítico obtenido para la casilla inicial u0.</font>
</p>

```python
# (I - A)u = b
I = np.eye(7)
M = I - A
vr = np.linalg.solve(M, b) # Valor real
print(f"Probabilidad Analítica: {vr[0]}")
```

<p><font size="3">
  Simulación de Monte Carlo<br><br>
  Definimos la lógica del experimento para simular el comportamiento aleatorio del ratón en el laberinto:
</font></p>

<p>
  <font size="3" color="green"><b>def s1():</b></font>
  <font size="3"> Función que simula un único recorrido completo del ratón.</font><br>
  
  <font size="3" color="green"><b>c = {...}</b></font>
  <font size="3"> Diccionario que mapea las conexiones y caminos disponibles desde cada casilla.</font><br>
  
  <font size="3" color="green"><b>pos = 0</b></font>
  <font size="3"> Define la casilla inicial (u0) donde el ratón comienza el juego.</font><br>
  
  <font size="3" color="green"><b>while pos not in [7, 8]:</b></font>
  <font size="3"> Ciclo que mantiene al ratón moviéndose hasta caer en la comida (7) o la trampa (8).</font><br>
  
  <font size="3" color="green"><b>pos = random.choice(c[pos])</b></font>
  <font size="3"> Selecciona de manera uniforme y al azar la siguiente casilla disponible.</font><br>
  
  <font size="3" color="green"><b>return 1 if pos == 7 else 0</b></font>
  <font size="3"> Devuelve 1 (éxito) si llegó a la comida, o 0 (fracaso) si cayó en la trampa.</font>
</p>

```python
# Simulacion
def s1():
    c = { # Caminos posibles
        0: [1, 2],
        1: [0, 3, 7],
        2: [0, 3, 8],
        3: [1, 2, 4, 5],
        4: [3, 6, 7],
        5: [3, 6, 8],
        6: [4, 5]
    }

    pos = 0 # Inicio
    while pos not in [7, 8]:  # Mientras no termine
        pos = random.choice(c[pos])  # Movimiento aleatorio
    return 1 if pos == 7 else 0
```

<p><font size="3">
  Repetimos la simulación una gran cantidad de veces para obtener la probabilidad por frecuencias:
</font></p>

<p>
  <font size="3" color="green"><b>n = 100000</b></font>
  <font size="3"> Número total de iteraciones o viajes independientes que realizará el ratón.</font><br>
  
  <font size="3" color="green"><b>resultados = np.array([...])</b></font>
  <font size="3"> Almacena en un arreglo de NumPy el resultado (0 o 1) de cada una de las n simulaciones.</font><br>
  
  <font size="3" color="green"><b>p1 = np.mean(resultados)</b></font>
  <font size="3"> Calcula la proporción de victorias obtenidas dividiendo los éxitos entre el total de intentos.</font>
  
  <font size="3" color="green"><b>print("Error:", abs(vr[0] - p1))</b></font>
  <font size="3"> Resta el resultado de la simulación al valor real y obtiene su valor absoluto para medir la precisión del experimento.</font>

</p>

```python
# Ejecucion
n = 100000  # Numero de simulaciones
resultados = np.array([s1() for _ in range(n)])

p1 = np.mean(resultados)
print(f"Probabilidad Simulada: {p1}")
print("Error:", abs(vr[0] - p1))
```

<h1><font>¿Cuál es el número promedio de tiradas necesarias para terminar el juego de serpientes y escaleras en el tablero de la imagen adjunta?</font></h1>
<p><font size="3"><b>Planteamiento Analitico</b><br><br>
  Modelamos el juego como una Cadena de Markov en tiempo discreto con espacio de estados <i>S = {0, 1, ..., 20}</i>.<br>El estado <b>20</b> es nuestro unico estado absorbente.<br><br>
  El dado es de 6 caras, por lo que la probabilidad de transicion al tirar el dado es de <b>1/6</b> para avanzar de 1 a 6 casillas. Si el tiro supera el tablero, el jugador permanece en su casilla actual. Las serpientes y escaleras modifican el estado de destino inmediato de la siguiente forma:
</font></p><br>
<p><font size="3">Estados absorbentes: Casilla 20 (Meta).</font></p>
<p><font size="3">Estados transitorios: Casillas del 0 al 19. Asi planteamos las reglas de transicion para construir la matriz del sistema:</font></p>

<p>
  <font size="3" color="green">Escaleras:</font><font size="3"> Si se cae en la casilla 3 se avanza a la 11; si se cae en la 15 se avanza a la 19.</font><br>
  <font size="3" color="red">Serpientes:</font><font size="3"> Si se cae en la casilla 13 se retrocede a la 4; si se cae en la 17 se retrocede a la 10.</font>
</p><br>

<p><font size="3">Solucion:<br>
  Para hallar el numero esperado de tiradas desde el inicio (casilla 0), definimos la matriz <i>A</i> de tamano 20x20 que contiene unicamente las probabilidades de transicion entre las casillas transitorias (del 0 al 19).<br><br>
  Cada paso se plantea con la ecuacion de balance general para el tiempo esperado de absorcion: <font size="3" color="brown">vr<sub>i</sub> = 1 + &sum; P<sub>i,j</sub> vr<sub>j</sub></font>.<br>
  Desglosando el sistema para las primeras casillas e incorporando los saltos directos del tablero, obtenemos las siguientes ecuaciones lineales:
</font></p>

<p>
  <font size="3" color="orange">vr<sub>0</sub> = 1 + 1/6 (vr<sub>1</sub> + vr<sub>2</sub> + vr<sub>11</sub> + vr<sub>4</sub> + vr<sub>5</sub> + vr<sub>6</sub>)</font><br>
  <font size="3" color="green">vr<sub>1</sub> = 1 + 1/6 (vr<sub>2</sub> + vr<sub>11</sub> + vr<sub>4</sub> + vr<sub>5</sub> + vr<sub>6</sub> + vr<sub>7</sub>)</font><br>
  <font size="3" color="blue">vr<sub>2</sub> = 1 + 1/6 (vr<sub>11</sub> + vr<sub>4</sub> + vr<sub>5</sub> + vr<sub>6</sub> + vr<sub>7</sub> + vr<sub>8</sub>)</font><br>
  <font size="3" color="brown">vr<sub>4</sub> = 1 + 1/6 (vr<sub>5</sub> + vr<sub>6</sub> + vr<sub>7</sub> + vr<sub>8</sub> + vr<sub>9</sub> + vr<sub>10</sub>)</font><br>
  <font size="3" color="red">vr<sub>5</sub> = 1 + 1/6 (vr<sub>6</sub> + vr<sub>7</sub> + vr<sub>8</sub> + vr<sub>9</sub> + vr<sub>10</sub> + vr<sub>11</sub>)</font><br>
  <font size="3" color="purple">vr<sub>6</sub> = 1 + 1/6 (vr<sub>7</sub> + vr<sub>8</sub> + vr<sub>9</sub> + vr<sub>10</sub> + vr<sub>11</sub> + vr<sub>12</sub>)</font>
</p>

<p><font size="3">
  Aplicando el metodo de reduccion por diferencias finitas, restamos ecuaciones consecutivas para simplificar algebraicamente la dependencia del sistema:<br><br>
  <font size="3" color="green">vr<sub>1</sub> - vr<sub>0</sub> = 1/6 (vr<sub>7</sub> - vr<sub>1</sub>) &rArr; vr<sub>7</sub> = 7 vr<sub>1</sub> - 6 vr<sub>0</sub></font><br>
  <font size="3" color="blue">vr<sub>2</sub> - vr<sub>1</sub> = 1/6 (vr<sub>8</sub> - vr<sub>2</sub>) &rArr; vr<sub>8</sub> = 7 vr<sub>2</sub> - 6 vr<sub>1</sub></font><br><br>
  Extendiendo este analisis de diferencias a lo largo del tablero y aplicando las condiciones de frontera impuestas por los bucles de retroceso (casilla 13 regresa a la 4, casilla 17 regresa a la 10 y casilla 15 salta a la 19), se plantea de forma analitica la matriz del sistema <font size="3" color="purple">M = I - A</font>.<br><br>
  Al efectuar la inversion matricial para hallar <font size="3" color="brown">M<sup>-1</sup> = (I - A)<sup>-1</sup></font> por el metodo de eliminacion lineal de Gauss-Jordan, el valor exacto de la primera fila correspondiente a la casilla de inicio se reduce a la siguiente fraccion algebraica:
</font></p>

<p><font size="3">
  <font size="3" color="red">vr<sub>0</sub> = 307 / 42</font>
</font></p>

<p><font size="3">
  <font color="red">vr<sub>0</sub> = 7.3106</font><br>
  Haciendo la division final de la fraccion obtenida del sistema, encontramos que el numero promedio de tiros para llegar a la meta es exactamente 7.3106.
</font></p>

<p><font size="3">
  <b>Proceso en Python</b><br>
  Construimos la matriz de probabilidades para el juego de serpientes y escaleras inicializando el espacio de estados y aplicando las reglas del tablero:
</font></p>

<p>
  <font size="3" color="brown"><b>A = np.zeros((20, 20))</b></font>
  <font size="3"> Crea una matriz vacía de 20x20 para almacenar las probabilidades de los estados transitorios.</font><br>
  
  <font size="3" color="brown"><b>for i in range(20):</b></font>
  <font size="3"> Recorre cada una de las casillas transitables del tablero (de la 0 a la 19).</font><br>
  
  <font size="3" color="brown"><b>for d in range(1, 7):</b></font>
  <font size="3"> Simula las 6 posibles caras del dado para calcular el avance desde la posición actual.</font><br>
  
  <font size="3" color="brown"><b>sig = i + d</b></font>
  <font size="3"> Calcula la casilla de posible destino sumando el valor obtenido en el dado.</font><br>
  
  <font size="3" color="brown"><b>if sig &lt; 20:</b></font>
  <font size="3"> Condición que valida que el movimiento no supere la meta (estado absorbente).</font><br>
  
  <font size="3" color="brown"><b>if sig == 3: sig = 11 ...</b></font>
  <font size="3"> Modifica la posición final si el jugador cae en una casilla con escalera o serpiente.</font><br>
  
  <font size="3" color="brown"><b>A[i, sig] += 1/6</b></font>
  <font size="3"> Acumula la probabilidad de 1/6 en la transición correspondiente de la matriz.</font>
</p>

```python
# Matriz A de Probabilidades de transicion
A = np.zeros((20, 20))

for i in range(20):
    for d in range(1, 7):
        sig = i + d
        if sig < 20:
            # Casillas especiales
            if sig == 3: sig = 11
            elif sig == 13: sig = 4
            elif sig == 15: sig = 19
            elif sig == 17: sig = 10
            A[i, sig] += 1/6
```

```python
print("A")
ver(A)
```

<p><font size="3">
  Planteamos la ecuación matricial para encontrar el número promedio de tiros necesarios antes de llegar a la meta:
</font></p>

<p>
  <font size="3" color="brown"><b>I = np.eye(20)</b></font>
  <font size="3"> Define una matriz identidad de 20x20 compatible con el tamaño del tablero.</font><br>
  
  <font size="3" color="brown"><b>M = I - A</b></font>
  <font size="3"> Resta la matriz de transición a la identidad para aislar matemáticamente el sistema.</font><br>
  
  <font size="3" color="brown"><b>vr = np.sum(np.linalg.inv(M), axis=1)</b></font>
  <font size="3"> Invierte la matriz M y suma sus filas para obtener el tiempo promedio de absorción de cada estado.</font><br>
  
  <font size="3" color="brown"><b>print(...)</b></font>
  <font size="3"> Muestra en la pantalla el valor analítico exacto partiendo desde la casilla inicial 0.</font>
</p>

```python
# (I - A)u = 1: Resolucion del sistema fundamental
I = np.eye(20)
M = I - A
vr = np.sum(np.linalg.inv(M), axis=1) # Tiempos esperados de absorcion
print(f"Valor real (Analitico): {vr[0]}")
```

<p><font size="3">Simulación Monte Carlo del Juego<br><br>
  Creamos la función para reproducir las reglas de un juego individual usando tiradas de dados aleatorias:
</font></p>

<p>
  <font size="3" color="brown"><b>def s2():</b></font>
  <font size="3"> Define la función para simular una partida completa de serpientes y escaleras.</font><br>
  
  <font size="3" color="brown"><b>pos = 0</b></font>
  <font size="3"> Establece la casilla de inicio del jugador en la posición cero.</font><br>
  
  <font size="3" color="brown"><b>t = 0</b></font>
  <font size="3"> Inicializa el contador que acumulará el número de turnos o tiros realizados.</font><br>
  
  <font size="3" color="brown"><b>while pos &lt; 20:</b></font>
  <font size="3"> Mantiene el ciclo activo mientras el jugador no alcance o supere la casilla 20.</font><br>
  
  <font size="3" color="brown"><b>t += 1</b></font>
  <font size="3"> Registra y suma un tiro de dado al contador en cada ciclo.</font><br>
  
  <font size="3" color="brown"><b>pos += random.randint(1, 6)</b></font>
  <font size="3"> Genera de forma uniforme un avance aleatorio entre 1 y 6 posiciones.</font><br>
  
  <font size="3" color="brown"><b>if pos &lt; 20:</b></font>
  <font size="3"> Valida que el jugador siga dentro del tablero para aplicar los efectos de las casillas.</font><br>
  
  <font size="3" color="brown"><b>if pos == 3: pos = 11 ...</b></font>
  <font size="3"> Evalúa si la posición coincide con una penalización o recompensa para reubicar al jugador.</font><br>
  
  <font size="3" color="brown"><b>return t</b></font>
  <font size="3"> Devuelve el total de tiradas que tomó terminar esa partida específica.</font>
</p>

```python
# Simulacion
def s2():
    pos = 0
    t = 0 # Contador de tiradas
    while pos < 20:
        t += 1
        pos += random.randint(1, 6) # Tiro de dado
        if pos < 20:
            # Evaluacion de serpientes y escaleras en el movimiento
            if pos == 3: pos = 11
            elif pos == 13: pos = 4
            elif pos == 15: pos = 19
            elif pos == 17: pos = 10
    return t
```

<p><font size="3">
  Repetimos la simulación una gran cantidad de veces para obtener la probabilidad por frecuencias:
</font></p>

<p>
  <font size="3" color="brown"><b>n = 100000</b></font>
  <font size="3"> Define el número total de juegos independientes que se van a simular.</font><br>
  
  <font size="3" color="brown"><b>resultados = np.array([...])</b></font>
  <font size="3"> Ejecuta las n partidas y guarda la cantidad de tiros de cada una en un arreglo.</font><br>
  
  <font size="3" color="brown"><b>p2 = np.mean(resultados)</b></font>
  <font size="3"> Calcula el promedio aritmético de tiros de todas las simulaciones corridas.</font><br>
  
  <font size="3" color="brown"><b>print(...)</b></font>
  <font size="3"> Imprime el resultado estimado por la simulación Monte Carlo y la diferencia absoluta contra el valor analítico.</font>
</p>

```python
# Ejecucion
n = 100000 # Numero de simulaciones
resultados = np.array([s2() for _ in range(n)])

p2 = np.mean(resultados) # Media de los pasos simulados
print(f"Probabilidad Simulada: {p2}")
print("Error:", abs(vr[0] - p2))
```
